Cada línea con contenido tiene tres campos separados por punto y coma:

Nombre ; correo ; programa

Los tres programas que existen en Bridge SA son Data Science, IA Engineering y Ciberseguridad. Los correos son las cuentas que la escuela asigna a cada participante, todas del dominio bridge.example.

>  La forma de lo que entregas

altas_limpias = [
    ["Alba Requena", "alba.requena@bridge.example", "Data Science"],
    ...
]

descartes = [
    ["la linea original, tal cual estaba", "email_invalido"],
    ...
]

*Guardar la línea original en el descarte (no la ya limpia) es lo que permite contestarle a Operations con la línea exacta que llegó. Es la mitad del encargo.*

 <span style = "color:red;"> " ".join(texto.split()) </span>deja exactamente un espacio entre palabras y quita los de los bordes, en una sola línea. Y "".join(texto.split()) los quita todos, que es lo que necesita un correo. Son el atajo clásico para limpiar algo escrito a mano.

```python
# Bridge SA · Operations — altas de la cohorte de marzo
#
# ALTAS_TEXTO ya esta cargado con el listado, tal cual.
# Cada linea util tiene la forma:  Nombre ; correo ; programa
#
# Deja preparadas estas tres cosas:
#   altas_limpias -> lista de listas [nombre, email, programa]
#   descartes     -> lista de listas [linea_original, motivo]
#   artefacto     -> el paquete que se lleva el Sprint 2
#
# Recuerda: linea en blanco se salta (no es descarte); correo sin @ o que
# no acabe en @bridge.example es descarte "email_invalido"; correo ya
# visto es descarte "duplicado". Y primero se limpia, despues se decide.
#
# Trucos: " ".join(t.split()) deja un solo espacio entre palabras;
#         "".join(t.split()) los quita todos;  t.title() pone la inicial
#         de cada palabra en mayuscula.

ALTAS_TEXTO = """  Alba   Requena ;  ALBA.REQUENA@bridge.example ; Data Science

sergio molina ; sergio.molina@bridge.example ; IA Engineering
Nuria Bel ; nuria.bel.bridge.example ; Data Science
CARLA VIDAL ; carla.vidal @Bridge.Example ; Ciberseguridad
  Iván  Tejero ; ivan.tejero@BRIDGE.example ;  Data  Science 

Marta Sanchís ; MARTA.SANCHIS@BRIDGE.EXAMPLE ; IA Engineering
SERGIO   MOLINA ;  Sergio.Molina@bridge.example  ; IA Engineering
Adrián Gascó ; adrian.gasco@bridge.example ; Ciberseguridad
  Noa Ferrán;noa.ferran@bridge.example;Data Science
Óscar Ibáñez ; oscar.ibanez@bridge.example ; IA Engineering 
elena  cortés ; ELENA.CORTES@bridge.example ; Data Science"""

altas_limpias = []
descartes = []
vistos = []

for linea in ALTAS_TEXTO.split("\n"): # la lista entera, dividela en saltos de linea
    if not linea.strip():             # si la linea no está vacía, continua ejecutando el bucle de más abajo
        continue
    
    nombre, correo, programa = linea.split(";") # dividimos lo que recorre linea con ";" y diseccionamos por nombre, correo y programa
    nombre = " ".join(nombre.split()).title()   # en nombre, primera letra de cada palabra en mayuscula, eliminamos espacios y separamos pabaras con un espacio
    correo = "".join(correo.split()).lower()    # en correo, ponemos todo en minuscula, eliminamos espacios y queda todo pegadito
    programa = " ".join(programa.split()).title()   # en programa, igual que en nombre
    
    # si los ultimos 15 caracteres de correo no terminan en @bridge.example o no hay "@", añadimos LINEA entera en descartes con motivo "email invalido"
    if correo[-15:] != "@bridge.example" or "@" not in correo:
        descartes.append((linea, "email_invalido"))
        continue
    
    # si hay un correo en VISTOS, lo añadimos a DESCARTES con motivo "duplicado"
    if correo in vistos:
        descartes.append((linea, "duplicado"))
        continue

    # si el correo no cumple las condiciones previas de descarte, añade ese correo en VISTOS y toda la info de LINEA en ALTAS_LIMPIAS
    vistos.append(correo)
    altas_limpias.append([nombre,correo,programa])

artefacto = {
    "artefacto": "altas_limpias",
    "fuente": "Operations & Cohortes",           # <- de donde salen estos datos
    "columnas": ["nombre", "email", "programa"],
    "altas": altas_limpias,
    "descartes": descartes,
}

print("Altas validas:", len(altas_limpias))
print("Descartes:", len(descartes))
```
