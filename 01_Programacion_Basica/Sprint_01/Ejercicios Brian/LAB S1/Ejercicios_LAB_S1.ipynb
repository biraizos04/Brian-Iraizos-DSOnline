#### Imprime los números del 1 al 10 saltando el 3, el 4 y el 9. Hazlo de dos maneras distintas (por ejemplo, una con not in y otra con continue).

```python
#Usando continue devuelve numeros por linea
exceptos = [3,4,9]
numeros=[]
for i in range(1,11):
    if i == 3:
        continue
    if i == 4:
        continue
    if i == 9:
        continue
    print(i)

#Usando el continue not in devuelve numeros en lista numeros[]

for i in range(1,11):
    if i not in exceptos:
        numeros.append(i)
print(numeros)

# Por qué funciona
# Las dos maneras dan la misma salida pero se leen distinto: con not in dices qué quieres imprimir, y con continue dices qué quieres saltar.
# continue gana cuando hay varias condiciones de descarte, porque evita anidar if dentro de if y deja el caso normal sin sangrar.
# Que existan varias formas correctas es normal: se elige por legibilidad.
```

#### Ya filtraste los "VW" con un for; ahora hazlo con un bucle while, recorriendo la lista por índice. Partiendo otra vez de coches = ["VW", "Audi", "Renault", "VW", "BMW"], construye en resultado una lista nueva con todos los coches que no sean "VW" y muéstrala.

```python
coches = ["VW", "Audi", "Renault", "VW", "BMW"]
resultado=[]
i = 0

while i < len(coches):
    if coches[i] != "VW":
        resultado.append(coches[i])
    i = i + 1
print(resultado)

# Por qué funciona
# El mismo filtro con while te obliga a llevar el índice a mano, y ahí está la lección: si olvidas i = i + 1, 
# la condición nunca cambia y el bucle no termina nunca. El for del ejercicio anterior avanza el índice por ti, 
# y por eso es lo que se usa para recorrer una colección entera; el while se reserva para cuando la parada no depende de la longitud.
```

##### Pequeño juego: el número secreto es secreto = 4 y tienes vidas = 2. Los intentos podrían pedirse con input(); aquí van en una lista fija, intentos = [3, 4] (cámbiala para probar). Recórrelos con un while mientras queden vidas: si aciertas, imprime "You win" y sal con break; si fallas, resta una vida. Si te quedas sin vidas, imprime "You lose".

Por partes:

    1 - Escribe las tres variables de partida y un índice i = 0 para saber por qué intento vas.
    2 - El while sigue mientras queden vidas y queden intentos en la lista.
    3 - Dentro, coge el intento de esa vuelta (intentos[i]) y avanza i.
    4 - Si el intento es el secreto, imprime y sal con break; si no, resta una vida.
    5 - Ya fuera del bucle, si las vidas han llegado a 0, has perdido.


```python
secreto = 4
vidas = 2
intentos = [3, 4]
i = 0

while i < len(intentos) and vidas > 0:
    intento_user = intentos[i]
    i = i + 1
    if intento_user == secreto:
        print("You win")
        break
    else:
        vidas = vidas - 1
if vidas == 0:
    print("You lose")
        
# Por qué funciona
# El break sale del bucle en cuanto aciertas, sin gastar los intentos que quedan: sin él seguirías comparando y
# podrías restar una vida después de haber ganado. Y la condición del while lleva dos partes (vidas > 0 and idx < len(intentos))
# porque hay dos motivos distintos para parar; olvidar el segundo te daría IndexError al pasarte del final de la lista.
```

#### El mismo programa del área del paralelogramo, con input, pero ahora tiene que admitir valores con decimales escritos en español, con coma, para la base y la altura. Las variables se llaman igual: base, altura y area. Tendrás que usar el método replace.

```python
# respuestas = ["8,5", "3,2"]

base = input("Escribe la base\n")
base = float(base.replace(",","."))
print(base)
altura = input("Escribe la altura\n")
altura = float(altura.replace(",","."))
print(altura)
area = base * altura
print(f"{float(area):.2f}")

# float() solo entiende el punto decimal, así que float("8,5") lanza ValueError en vez de interpretar la coma como haría un lector español. 
# Por eso primero se normaliza el texto y después se convierte: limpiar antes de convertir es el orden habitual con cualquier dato que venga de fuera. 
# El mismo problema reaparece al leer un CSV con decimales españoles.
```
